## Engineering Colleges in Andhra Pradesh

This project analyzes engineering colleges in Andhra Pradesh based on district, college type, branch, fees, NAAC grade, and placement percentage.

```python
import pandas as pd

df = pd.read_csv("AP_Engineering_Colleges.csv")

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
import matplotlib.pyplot as plt

df['District'].value_counts().plot(kind='bar')

plt.title("Engineering Colleges by District")
plt.xlabel("District")
plt.ylabel("Number of Colleges")

plt.show()
```

```python
df['Type'].value_counts().plot(
      kind='pie',
          autopct='%1.1f%%'
          )

plt.title("Government vs Private Colleges")

plt.show()
```

```python
df['Branch '].value_counts().plot(kind='bar')

plt.title("Branch Distribution")
plt.xlabel("Branch")
plt.ylabel("Count")

plt.show()
```

Insights and Findings

1. Most engineering colleges in the dataset are private institutions.

2. Guntur, Visakhapatnam, and Krishna districts contain multiple engineering colleges.

3. Computer Science Engineering (CSE) is one of the most popular branches.

4. College fees vary significantly among institutions.

5. Placement percentages differ across colleges, helping students compare opportunities.

Conclusion

The analysis of engineering colleges in Andhra Pradesh provides useful insights regarding college distribution, branch availability, fees, and placements. Students can use this information to compare institutions and make informed decisions about higher education.
